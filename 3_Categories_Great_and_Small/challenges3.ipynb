### 3. Bool as a Monoid with AND and OR  
**A monoid** is a set with a binary operation that is associative and has an identity element

(Bool, AND)
- Set: {True, False}
- Operation: AND
- Identity element: True
- Associativity: (a && b) && c = a && (b && c) for all a, b, c
- Identity law: a && True = True && a = a

(Bool, OR)
- Set: {True, False}
- Operation: OR
- Identity element: False
- Associativity: (a || b) || c = a || (b || c) for all a, b, c
- Identity law: a || False = False || a = a

Thus, (Bool, &&, True) and (Bool, ||, False) are both monoids

### 4. Bool Monoid with AND as a Category
We can view the monoid (Bool, AND, True) as a category with one object 
- Object: There is a single object, call it `•` (or just `Bool`)
- Morphisms: The morphisms correspond to the elements of the monoid: `True` and `False`
- Identity morphism: `True` acts as the identity morphism
- Composition: Morphism composition is given by the AND operation:
    - `True ∘ True = True`
    - `True ∘ False = False ∘ True = False`
    - `False ∘ False = False`


### Composition Table
| ∘     | True  | False |
|-------|-------|-------|
| True  | True  | False |
| False | False | False |

This satisfies the category axioms:
- Associativity: (a && b) && c = a && (b && c)
- Identity: True && a = a && True = a

### 5. Addition modulo 3 as a Category
We can view the monoid $(Z_3, +, 0)$ as a category with one object 
- Object: There is a single object, $Z_3$
- Morphisms: The morphisms correspond to the elements of $Z_3$: $0, 1, 2$ (representing the functions $x \mapsto x + 0$, $x \mapsto x + 1$, $x \mapsto x + 2$ modulo 3)
- Identity morphism: $0$ acts as the identity morphism
- Composition: Morphism composition is given by addition modulo 3:
    - $0 \circ 0 = 0$
    - $0 \circ 1 = 1 \circ 0 = 1$
    - $0 \circ 2 = 2 \circ 0 = 2$
    - $1 \circ 1 = 2$
    - $1 \circ 2 = 0$
    - $2 \circ 1 = 0$
    - $2 \circ 2 = 1$


This satisfies the category axioms:
- Associativity: $(a + b) + c \equiv a + (b + c) \pmod{3}$
- Identity: $0 + a = a + 0 = a$
