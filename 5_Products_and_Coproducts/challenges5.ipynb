### 2. What is a product of two objects in a poset? Hint: Use the universal construction.

A **poset** (partially ordered set) can be viewed as a category where:
- The **objects** are the elements of the set.
- A **morphism** exists from `a` to `b` if and only if `a ≤ b`. There is at most one morphism between any two objects.

The **universal construction of a product** of two objects `a` and `b` defines an object `p` with two properties:

1.  **Projections**: `p` must have morphisms to `a` and `b`. In a poset, this means:
    - `p ≤ a`
    - `p ≤ b`
    So, `p` must be a **lower bound** of `a` and `b`.

2.  **Universality**: For any *other* object `x` that is also a lower bound (`x ≤ a` and `x ≤ b`), there must be a unique morphism from `x` to `p`. In a poset, this means:
    - `x ≤ p`

Combining these, `p` is a lower bound that is greater than or equal to all other lower bounds.

**Example**: In the poset of integers with the usual `≤` relation, the product of 5 and 7 is `min(5, 7) = 5`.
- **Projections**: `5 ≤ 5` and `5 ≤ 7`.
- **Universality**: Any other number `x` such that `x ≤ 5` and `x ≤ 7` is also less than or equal to 5 (i.e., `x ≤ 5`).

### 3. What is a coproduct of two objects in a poset?

The **coproduct** is the dual of the product. In a poset, the coproduct of two objects `a` and `b` is their **least upper bound (LUB)**, also known as the **supremum** or **join**.

This is derived from the dual universal construction:

1.  **Injections**: A coproduct `c` must have morphisms coming *from* `a` and `b`. In a poset, this means:
    - `a ≤ c`
    - `b ≤ c`
    So, `c` must be an **upper bound**.

2.  **Universality**: For any *other* object `y` that is also an upper bound (`a ≤ y` and `b ≤ y`), there must be a unique morphism from `c` to `y`. In a poset, this means:
    - `c ≤ y`

Combining these, `c` is an upper bound that is less than or equal to all other upper bounds.

**Example**: In the poset of integers with the usual `≤` relation, the coproduct of 5 and 7 is `max(5, 7) = 7`.
- **Injections**: `5 ≤ 7` and `7 ≤ 7`.
- **Universality**: Any other number `y` such that `5 ≤ y` and `7 ≤ y` is also greater than or equal to 7 (i.e., `7 ≤ y`).

### 4. Implement the equivalent of Haskell Either as a generic type in
your favorite language (other than Haskell).

````python
class Either:
    def __init__(self, left=None, right=None):
        self.left = left
        self.right = right

    @staticmethod
    def make_left(value):
        return Either(left=value)

    @staticmethod
    def make_right(value):
        return Either(right=value)

    @property
    def is_left(self):
        return self.left is not None

    @property
    def is_right(self):
        return self.right is not None
````

````python
# Testing
el = Either.make_left(5)
print(el.is_left)
print(el.left)

er = Either.make_right("Hello")
print(er.is_left)
print(er.right)
````

### 5. Show that Either is a “better” coproduct than int equipped with two injections:
```cpp
int i(int n) { return n; }
int j(bool b) { return b? 0: 1; }
```
Hint: Define a function
`int m(Either const & e);`
that factorizes i and j.

````python
def i(n: int) -> int:
    return n

def j(b: bool) -> int:
    return 0 if b else 1

def m(e: Either) -> int:
    if e.is_left:
        return e.left
    return 0 if e.right else 1
````
